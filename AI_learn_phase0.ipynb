```python
import pandas as pd
# 1. Build a DataFrame with columns name, age, score:
#    Alice 25 88, Bob 30 92, Carol 22 75
# 2. Print just the score column
# 3. Print only the rows where score is above 80

data = {
    "name" : ["Alice", "Bob", "Carol"],
    "age" : [25, 30, 22],
    "score" : [88, 92, 75]
}
df = pd.DataFrame(data)

print(df["score"])
print(df[df['score'] > 80])

```

```python
# 1. Add a new column "passed" that is True where score >= 80, False otherwise
#    (hint: it's just the comparison from task 3, assigned to a new column)
# 2. Print only the names of the people who passed
#    (hint: filter the rows, then select the column — combine both skills)

import pandas as pd

data = {
    "name" : ["Alice", "Bob", "Carol"],
    "age" : [25, 30, 22],
    "score" : [88, 92, 75]
}
df = pd.DataFrame(data)

df["passed"] = df["score"] >= 80
print(df['name'][df["passed"]])
```

```python
# 1. Print summary stats for the numeric columns (hint: .describe())
# 2. Print the DataFrame sorted by score, highest first (hint: .sort_values)
# 3. What is the average age of the people who passed?
#    (combine today's skills: filter the rows, select the column, take the mean)

import pandas as pd

data = {
    "name" : ["Alice", "Bob", "Carol"],
    "age" : [25, 30, 22],
    "score" : [88, 92, 75]
}
df = pd.DataFrame(data)

df["passed"] = df["score"] >= 80
print(df.describe())

sort_df = df.sort_values(by = 'score', ascending=False)
print("")
print("=== Sort by score ===")
print(sort_df)

print("")
print("=== Everage age for passed ===")
print(df['age'][df['passed']].mean())
```
